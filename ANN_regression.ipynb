```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
data=pd.read_csv("Life Expectancy Data.csv")
data.head()
print(data.shape)
```

```python
#clean column
data.columns=data.columns.str.strip()
print(data.columns)
```

```python
data=data.dropna()
print(data.shape)
```

```python
print("Unique values in 'Status' before mapping:", data["Status"].astype(str).str.strip().unique())
data["Status"]=data["Status"].astype(str).str.strip().map({"Developing":0,"Developed":1})
print(data.Status)
```

```python
x=data.drop(["Life expectancy"],axis=1)
y=data["Life expectancy"]
```

```python
#converting category data in to number
x = pd.get_dummies(x, columns=["Country"], dtype=int)
```

```python
#train_test split
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
Scalar=StandardScaler()
x_train=Scalar.fit_transform(x_train)
x_test=Scalar.fit_transform(x_test)
```

```python
#ANN Model building
model=Sequential()
model.add(Dense(128,activation="relu",input_shape=(x_train.shape[1],)))
model.add(Dense(64,activation="relu"))
model.add(Dense(32,activation="relu"))
model.add(Dense(1,activation="linear"))
```

```python
#compling the neural network
model.compile(optimizer='adagrad',loss='mean_squared_error',metrics=['mae'])
```

```python
model.summary()
```

```python
history =model.fit(x_train,y_train,epochs=100,batch_size=32,validation_split=0.2)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
y_pred=model.predict(x_test)
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print("Mean Squared Error:",mse)
print("Mean Absolute Error:",mae)
print("R2 Score:",r2)
```
